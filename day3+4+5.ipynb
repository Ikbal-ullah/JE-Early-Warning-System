```python
import pandas as pd
#1. Load Dataset
df =pd.read_csv('assam_complete_historical_weather.csv')
#2. Sort the data
df=df.sort_values(by=['District','Date']).reset_index(drop=True)
#3.Cumulative Rainfall Features (The "Sum" over X days)
#7-Day: Captures flash flooding and immediate puddle formation
df['7_Day_Rain_Sum']=df.groupby('District')['Daily_Rain_mm'].transform(
    lambda x:x.rolling(window=7,min_periods=1).sum()
)
#14-Day: Captures sustained breeding ground formation
df['14_Day_Rain_Sum']=df.groupby('District')['Daily_Rain_mm'].transform(
    lambda x:x.rolling(window=14,min_periods=1).sum()
)
#30-Day: Captures macro-seasonal shifts (Monsoon onset)
df['30_Day_Rain_Sum']=df.groupby('District')['Daily_Rain_mm'].transform(
    lambda x:x.rolling(window=30,min_periods=1).sum()
)

#4. Average Temperature Feature (The "Mean" over X days)
#14-Day Avg: Dictates how fast larvae mature into adult mosquitoes
df['14_Day_Avg_Temp_C'] = df.groupby('District')['Max_Temp_C'].transform(
    lambda x: x.rolling(window=14, min_periods=1).mean()
)

# 5. Verify the math by isolating one district and looking at the new columns
kamrup_sample=df[df['District']=='Kamrup_Metropolitan'].tail(15)
print(kamrup_sample[['Date', 'Daily_Rain_mm', '7_Day_Rain_Sum', '14_Day_Avg_Temp_C']])
```
